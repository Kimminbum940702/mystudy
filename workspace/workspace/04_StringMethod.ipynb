len()함수는 인수로 지정된 문자열을 구성하는 문자의 개수를 얻어온다.

```python
string = 'We are the champions, My friends!'
print(len(string))
```

count() 메소드는 문자열에서 인수로 지정한 문자열의 출현 횟수를 얻어온다.  

```python
string = 'We are the champions, My friends!'
print(string.count('e'))
print(string.count('We'))
print(string.count('we'))
```

index() 메소드는 문자열에서 인수로 지정한 문자열이 최초로 출현하는 인덱스를 얻어온다.

```python
string = 'We are the champions, My friends!'
print(string.index('e'))
print(string.index('We'))
print(string.index('we')) # 찾으려는 문자열이 없으면 에러가 발생한다.
```

find() 메소드는 문자열에서 인수로 지정한 문자열이 최초로 출현하는 인덱스를 얻어온다.

```python
string = 'We are the champions, My friends!'
print(string.find('e'))
print(string.find('We'))
print(string.find('we')) # 찾으려는 문자열이 없으면 -1 리턴한다.
```

upper() 메소드는 문자열에서 인수로 지정한 문자열을 무조건 대문자로 변환한다.  
lower() 메소드는 문자열에서 인수로 지정한 문자열을 무조건 소문자로 변환한다.  
title() 메소드는 문자열에서 인수로 지정한 문자열의 첫 문자는 대문자로 나머지는 소문자로 변환한다. 

```python
string = 'We are the champions, My friends!'
print(string.upper())
print(string.lower())
print(string.title())
```

strip() 메소드는 문자열 앞,뒤의 불필요한 공백을 제거한다.  
lstrip() 메소드는 문자열 앞의 불필요한 공백을 제거한다.  
rstrip() 메소드는 문자열 뒤의 불필요한 공백을 제거한다.  

```python
string='   samsungAXI    '
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

replace() 메소드는 특정 문자열을 지정한 문자열로 일괄 치환한다.

```python
string='We are the champions,My friends!'
print(string.replace('e','2'))
```

join() 메소드는 인수로 지정된 문자열 사이에 특정 문자열을 삽입해서 연결한다.  

```python
string='samsungAXI'
print('/'.join(string))
print(' '.join(string))
print('\n'.join(string))
print('^^'.join(string))
```
