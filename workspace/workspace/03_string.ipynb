큰따옴표 또는 작은따옴표 3개를 연속해서 사용하면 여러줄 문자열을 만들 수 있다.

```python
memo = '''
 개울가에 올챙이 한마리
 꼬물꼬물 헤엄치다.
 앞다리가 쏘옥 뒷다리가 쑤욱
 팔딱팔딱 개구리됐네
'''
print(memo)
```

dedent() 함수는 여러 줄로 이루어진 문자열에서 공통된 들여쓰기를 제거할 때 사용한다.  
클래스의 docstring이나 들여쓰기가 포함된 html,sql등의 멀티라인 문자열을 깔끔하게 정렬할 때 주로 사용한다.  

```python
from textwrap import dedent
```

```python
memo = dedent("""
          개울가에 올챙이 한마리
          꼬물꼬물 헤엄치다.
          앞다리가 쏘옥 뒷다리가 쑤욱
          팔딱팔딱 개구리됐네
 """)  
print(memo)
```

# 인덱싱과 슬라이싱

인덱스는 '0'부텅 시작한다.  
'인덱싱': 문자열에서 특정 인덱스의 문자 1개를 얻어온다.=>[]에 인덱스를 지정하면 된다.    
'슬라이싱': 문자열에서 지정한 범위의 문자열을 얻어온다.=>[시작인덱스:끝인덱스] 형태로 범위를 지정한다.    

인덱싱

```python
memo = '030152-265123'
print(memo[7])
# 파이썬은 음수 인데싱이 가능하다.
print(memo[-1])
```

## 슬라이싱

[시작인덱스:끝인덱스]형태로 범위를 지정하며 시작인데스 부터 "끝인덱스 -1"번째 문자열을 얻어온다.

```python
memo = '030152-265123'
print(memo[0:6])
# 시작인덱스를 생략하면 처음부터 슬라이싱하고 끝인덱스를 생략하면 마지막까지 슬라이싱한다.
print(memo[:6])
print(memo[7:])
```
